```python
# 2. Persistence Prediction: Tomorrow's min temp = Today's min temp
# Using Basel as our test city
df['persistence_pred_temp'] = df['BASEL_temp_min']
```

```python
# 2. Recreate the target columns (Tomorrow's min temp)
df['target_future_min_temp'] = df[f'{city_prefix}_temp_min'].shift(-1)
df['frost_label'] = (df['target_future_min_temp'] <= 3.0).astype(int)
```

```python
# 3. Persistence Prediction: Tomorrow's min temp = Today's min temp
df['persistence_pred_temp'] = df[f'{city_prefix}_temp_min']
df['persistence_pred_label'] = (df['persistence_pred_temp'] <= 3.0).astype(int)
```

```python
# 4. Evaluate the predictions
# Drop the last row since it won't have a 'tomorrow' value due to the shift
clean_df = df.dropna(subset=['target_future_min_temp', 'persistence_pred_temp'])

cont_metrics = get_continuous_metrics(clean_df['target_future_min_temp'], clean_df['persistence_pred_temp'])
class_metrics = get_classification_metrics(clean_df['frost_label'], clean_df['persistence_pred_label'])
```

```python
print("--- Persistence Baseline Results ---")
print(f"RMSE: {cont_metrics['rmse']:.2f}°C, MAE: {cont_metrics['mae']:.2f}°C")
print(f"Precision: {class_metrics['precision']:.2f}, Recall: {class_metrics['recall']:.2f}")
print(f"F1: {class_metrics['f1']:.2f}, Peirce Skill: {class_metrics['peirce_skill']:.2f}")

# 5. Log the results to experiments/runs.csv
run_data = [
    datetime.now().strftime("%Y-%m-%d %H:%M"),
    "Persistence", 
    "None", 
    "Basel_Tier1",
    round(class_metrics['precision'], 3),
    round(class_metrics['recall'], 3),
    round(class_metrics['f1'], 3),
    round(class_metrics['peirce_skill'], 3),
    round(cont_metrics['rmse'], 3),
    round(cont_metrics['mae'], 3)
]

with open('../experiments/runs.csv', 'a', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(run_data)
print("\nSuccess: Baseline logged to experiments/runs.csv")
```
